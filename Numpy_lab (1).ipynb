```python
print("AP24110011070")
# Import NumPy
import numpy as np

# Create the first pair of similar sentences
sentence1 = "I love playing football"
sentence2 = "I enjoy playing football"

# Create the second pair of different sentences
sentence3 = "I love playing football"
sentence4 = "The weather is very cold today"

# Split the sentences into words
tokens1 = sentence1.split()

# Split the sentences into words
tokens2 = sentence2.split()

# Split the sentences into words
tokens3 = sentence3.split()

# Split the sentences into words
tokens4 = sentence4.split()

# Create a vocabulary for the first pair
vocab_similar = sorted(set(tokens1 + tokens2))

# Create a vocabulary for the second pair
vocab_different = sorted(set(tokens3 + tokens4))

# Create the bag-of-words vector for the first sentence
vector1 = np.array([tokens1.count(word) for word in vocab_similar])

# Create the bag-of-words vector for the second sentence
vector2 = np.array([tokens2.count(word) for word in vocab_similar])

# Create the bag-of-words vector for the third sentence
vector3 = np.array([tokens3.count(word) for word in vocab_different])

# Create the bag-of-words vector for the fourth sentence
vector4 = np.array([tokens4.count(word) for word in vocab_different])

# Calculate cosine similarity for the similar pair
similarity1 = np.dot(vector1, vector2) / (np.linalg.norm(vector1) * np.linalg.norm(vector2))

# Calculate cosine similarity for the different pair
similarity2 = np.dot(vector3, vector4) / (np.linalg.norm(vector3) * np.linalg.norm(vector4))

# Display the similar sentences
print("Similar sentences:")
print(sentence1)
print(sentence2)

# Display the cosine similarity for the similar pair
print("Cosine similarity:", similarity1)

# Display the different sentences
print("\nDifferent sentences:")
print(sentence3)
print(sentence4)

# Display the cosine similarity for the different pair
print("Cosine similarity:", similarity2)
```

```python
print("AP24110011070")
# Import the NumPy library
import numpy as np

# Create a paragraph containing four sentences
paragraph = "Machine learning is a powerful technology. Machine learning helps computers learn from data. Data is used to train machine learning models. These models can solve many real world problems."

# Tokenize the paragraph into individual words
tokens = paragraph.lower().replace(".", "").split()

# Build the vocabulary using unique words
vocab = sorted(set(tokens))

# Create a function to calculate the word-count vector
def bag_of_words_vector(tokens, vocab):

    # Count each word in the vocabulary
    vector = np.array([tokens.count(word) for word in vocab])

    # Return the word-count vector
    return vector

# Create the word-count vector
word_counts = bag_of_words_vector(tokens, vocab)

# Sort the indices by word frequency in descending order
top_indices = np.argsort(word_counts)[::-1]

# Select the top 3 most frequent word indices
top_3_indices = top_indices[:3]

# Display the top 3 most frequent words and their counts
print("Top 3 most frequent words:")

# Print each top word and its frequency
for index in top_3_indices:

    # Display the word and its count
    print(vocab[index], ":", word_counts[index])
```
