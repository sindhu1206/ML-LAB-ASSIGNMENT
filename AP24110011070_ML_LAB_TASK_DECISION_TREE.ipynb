```python
# import libraries
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split #AP24110011070
```

```python
# load the data
iris = load_iris()
X = iris.data
y = iris.target
```

```python
# split the data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)
```

```python
# train the model AP24110010015
clf = DecisionTreeClassifier()
clf.fit(X_train, y_train)
```

```python
# test the model
y_pred = clf.predict(X_test)
```

```python
# evaluate the model
from sklearn.metrics import accuracy_score
print("Accuracy:", accuracy_score(y_test, y_pred))  #AP24110010015
```

```python
import seaborn as sns

data = sns.load_dataset("iris")

data
```

```python
from sklearn.model_selection import train_test_split

# Split the iris dataset into training and testing data
train_data, test_data = train_test_split(
    data,
    test_size=0.3,
    random_state=123
)

print("Training data:")
print(train_data)

print("\nTesting data:")
print(test_data)
```

```python
from sklearn.tree import DecisionTreeClassifier

# Create the model
fit = DecisionTreeClassifier(random_state=123)

# Separate features and target
X_train = train_data.drop("species", axis=1)
y_train = train_data["species"]

# Train the model
fit.fit(X_train, y_train)

print("Decision Tree model trained successfully.")
```

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))

plot_tree(
    fit,
    feature_names=X_train.columns,
    class_names=fit.classes_,
    filled=True
)

plt.title("Decision Tree")
plt.show()
```

```python
# Make predictions on the test data
predictions = fit.predict(X_test)

print("Predictions:")
print(predictions)    #AP24110010015
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, predictions)

print("Accuracy:", accuracy)
```

```python
# Clean Python version of the Iris Decision Tree AP24110010015

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# 1. Load the Iris dataset
import seaborn as sns

data = sns.load_dataset("iris")

# 2. Separate input and output
X = data.drop("species", axis=1)
y = data["species"]

# 3. Split into training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,
    random_state=123,
    stratify=y
)

# 4. Create and train Decision Tree
fit = DecisionTreeClassifier(random_state=123)
fit.fit(X_train, y_train)

# 5. Make predictions
predictions = fit.predict(X_test)

# 6. Calculate accuracy
accuracy = accuracy_score(y_test, predictions)

print("Accuracy:", accuracy)
```
